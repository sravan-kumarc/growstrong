```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate

user_query = "write a Simple python program to return a+b \n"

SYSTEM_PROMPT = """You are a helpful assistant that translates natural language to python code.
You will be given a natural language query and you will return a python code snippet that solves the query. 
The code should be well-formatted and follow best practices. Do not include any explanations or comments in the code. Only return the code snippet.
"""
llm = ChatOllama(model="llama3.2",
                temperature=0.7,
                top_p=0.9,
                max_tokens=2000)

llm_prompt = ChatPromptTemplate.from_messages([
    ("system", SYSTEM_PROMPT),
    ("user", "{user_query}")
])

response = llm.invoke(user_query)

#pprint(response.content)

for chunks in llm.stream(user_query):
    print(chunks.content, end="", flush=True)
```

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

user_query = "write a kubernetes imperative commands for Pod, Deployment, Service, ServiceAccount  \n"

SYSTEM_PROMPT = """You are a helpful assistant that translates natural language to python code.
You will be given a natural language query and you will return a python code snippet that solves the query. 
The code should be well-formatted and follow best practices. Do not include any explanations or comments in the code. Only return the code snippet.
"""
llm = ChatOllama(model="llama3.2",
                temperature=0.7,
                top_p=0.9,
                max_tokens=2000)

llm_prompt = ChatPromptTemplate.from_messages([
    ("system", SYSTEM_PROMPT),
    ("user", "{user_query}")
])

chain = llm_prompt | llm | StrOutputParser()

for chunks in chain.stream(user_query):
    print(chunks, end="", flush=True)
    
```
