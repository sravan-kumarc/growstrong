## Streaming a Natural-Language Response

Use `llm.stream()` to print a generated answer incrementally instead of waiting for the complete response.

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(model="llama3.2",
                temperature=0.7,
                top_p=0.9,
                max_tokens=2000)

user_query = "write a Simple python program to print 'Hello, World!' \n"

response = llm.invoke(user_query)

for chunks in llm.stream(user_query):
    print(chunks.content, end="", flush=True)
```

## Building a Prompt Template

Define a reusable system instruction and user-message placeholder with `ChatPromptTemplate`.

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate

user_query = "write a Simple python program to return a+b \n"

SYSTEM_PROMPT = """You are a helpful assistant that translates natural language to python code.
You will be given a natural language query and you will return a python code snippet that solves the query. 
The code should be well-formatted and follow best practices. Do not include any explanations or comments in the code. Only return the code snippet.
"""
llm = ChatOllama(model="llama3.2",
                temperature=0.7,
                top_p=0.9,
                max_tokens=2000)

llm_prompt = ChatPromptTemplate.from_messages([
    ("system", SYSTEM_PROMPT),
    ("user", "{user_query}")
])

response = llm.invoke(user_query)

#pprint(response.content)

for chunks in llm.stream(user_query):
    print(chunks.content, end="", flush=True)
```

## Translating Requests into Python

Combine a detailed system prompt with a user query, then stream the model's code-oriented response.

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

user_query = "write a kubernetes imperative commands for Pod, Deployment, Service, ServiceAccount  \n"

SYSTEM_PROMPT = """You are a helpful assistant that translates natural language to python code.
You will be given a natural language query and you will return a python code snippet that solves the query. 
The code should be well-formatted and follow best practices. Do not include any explanations or comments in the code. Only return the code snippet.
"""
llm = ChatOllama(model="llama3.2",
                temperature=0.7,
                top_p=0.9,
                max_tokens=2000)

llm_prompt = ChatPromptTemplate.from_messages([
    ("system", SYSTEM_PROMPT),
    ("user", "{user_query}")
])

chain = llm_prompt | llm | StrOutputParser()

for chunks in chain.stream(user_query):
    print(chunks, end="", flush=True)
    
```

## Composing a LangChain Pipeline

Connect the prompt, Ollama model, and `StrOutputParser` with the runnable pipe operator, then stream the parsed text.

```python
#without Runnables
import asyncio
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatOllama(model="llama3.2",
                temperature=0.7)

parser = StrOutputParser()

SYSTEM_PROMPT = """You are a helpful assistant. Provide concise, high-quality answers."""

llm_prompt = ChatPromptTemplate.from_messages([
    ("system", SYSTEM_PROMPT),
    ("user", "{user_query}")
])

chain = llm_prompt | llm | parser

##Actual Logic to run multiple tasks concurrently
async def tasks_run(task_id:int, user_query:str):
    print(f"Task {task_id} started.")
    async for chunks in chain.astream({"user_query": user_query}):
        print(f"Task {task_id}: {chunks}", end="", flush=True)
    print(f"\nTask {task_id} completed.")


async def main():
    user_queries = [
        "write a kubernetes imperative commands for ServiceAccount",
        "write a simple python program to return a+b",
        "what is 22+44+0"
    ]

    tasks = [tasks_run(i + 1, query) for i, query in enumerate(user_queries)]
    
    # Run all tasks concurrently
    await asyncio.gather(*tasks)

# 6. Run inside Jupyter Notebook (.ipynb)
await main()
```

## Running Requests Concurrently

Use `asyncio.gather()` with `chain.astream()` to process several independent user queries at the same time.

```python
# used runnable and chain to run multiple tasks concurrently

from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel


llm = ChatOllama(model="llama3.2",
                temperature=0.7)
parser = StrOutputParser()

query_one = "write a kubernetes imperative commands for ServiceAccount"
response_for_one = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. Provide concise, high-quality answers."),
    ("user", "{user_query1}")
])
query_one_chain = response_for_one | llm | parser


query_two = "write a simple python program to return a+b"
response_for_two = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. Provide concise, high-quality answers."),
    ("user", "{user_query2}")
])
query_two_chain = response_for_two | llm | parser

parallel_chain = RunnableParallel(
    q1=query_one_chain, 
    q2=query_two_chain)

if __name__ == "__main__":
    user_queries = {
        "user_query1": query_one,
        "user_query2": query_two
    }
    results = parallel_chain.invoke(user_queries)
    print("Results for Query 1:", results["q1"])
    print("Results for Query 2:", results["q2"])
```

## Branching with RunnableParallel

Create two independent prompt-model-parser chains and execute them together with `RunnableParallel`.

```python
%pip install -q langchain-community

from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.output_parsers import StrOutputParser
from langchain_ollama import ChatOllama
from langchain_core.chat_history import InMemoryChatMessageHistory

llm = ChatOllama(model="llama3.2", temperature=0.7)
parser = StrOutputParser()
history = InMemoryChatMessageHistory()

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. Provide concise, high-quality answers."),
    MessagesPlaceholder(variable_name="chat_history"),
    ("user", "{user_query}")
])

chain = prompt_template | llm | parser

def run_chat_with_history(user_query):

    # Prepare the input for the chain
    input_data = {
        "user_query": user_query,
        "chat_history": history.messages
    }

    # Invoke the chain
    response = chain.invoke(input_data)

    # Add the user query and assistant response to the chat history
    history.add_user_message(user_query)
    history.add_ai_message(response)

    return response

run_chat_with_history("Hello, What is sum of 22+5?")
run_chat_with_history("what are the integers we used for sum calculation?")


```

## Defining a LangChain Tool

Decorate a typed Python function with `@tool` so LangChain can expose its name, description, and argument schema.

```python
from langchain_core.tools import Tool, tool

@tool
def add_numbers(a: int, b: int) -> int:
    """Adds two numbers together."""
    return a + b

print(add_numbers.name)  # Output: add_numbers
print(add_numbers.description)  # Output: Adds two numbers together.
print(add_numbers.args)  # Output: {'a': {'type': 'int'}, 'b': {'type': 'int'}}
```

## Customizing a Tool Name

Give the decorated addition function a custom tool name while keeping its typed arguments and description.

```python
from langchain_core.tools import Tool, tool

@tool("MyIntsAreAdding")
def add_numbers(a: int, b: int) -> int:
    """Adds two numbers together."""
    return a + b

print(add_numbers.name)  # Output: MyIntsAreAdding
print(add_numbers.description)  # Output: Adds two numbers together.
print(add_numbers.args)  # Output: {'a': {'type': 'int'}, 'b': {'type': 'int'}}
```

## Binding Multiple Tools to the Model

Register addition and multiplication tools with the chat model, then inspect the tool-call response produced for a user request.

```python
from langchain_ollama import ChatOllama
from langchain_core.tools import Tool, tool

@tool("MyIntsAreAdding")
def add_numbers(a: int, b: int) -> int:
    """Adds two numbers together."""
    return a + b

@tool
def multiply_numbers(a: int, b: int) -> int:
    """Multiplies two numbers together."""
    return a * b


llm = ChatOllama(model="llama3.2",temperature=0)
tools = [add_numbers, multiply_numbers]
tools_binding = llm.bind_tools(tools)

prompt_for_mul = "What is 15 multiplied by 4?"
response = tools_binding.invoke(prompt_for_mul)

prompt_for_add = "What is 15 plus 4?"
response = tools_binding.invoke(prompt_for_add)


print("-"*30)
print(type(response))
print(response.tool_calls)
```
