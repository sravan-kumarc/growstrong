```python
from langchain_ollama import ChatOllama
from pprint import pprint

user_prompt = input("How may i help you? \n")
llm = ChatOllama(model="llama3.2", 
                temperature=0.7,
                top_p=0.9,
                max_tokens=2000)

response = llm.invoke(user_prompt)
pprint(response.content)
```
